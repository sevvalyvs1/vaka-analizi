## Bölüm 5 — Yapay zekâ kullanım günlüğü

**1. Araçlar ve kullanım biçimi**
- **Claude (sohbet):** Analiz planı, kodların büyük kısmı, ara sonuçların yorumlanması ve taslak raporun eleştirisi. Kodları adım adım çalıştırdım; her çıktıyı beklenen değerle karşılaştırıp doğruladıktan sonra bir sonraki adıma geçtim ve önerilerin bir kısmını sorgulayarak değiştirdim.
- **Groq, `openai/gpt-oss-20b`:** 200 geri bildirimin etiketlenmesi (3.1, bonus) ve aksiyon önerileri (3.3). Gemini API sürekli 503 (aşırı yoğunluk) hatası verdiği için Groq'a geçtim; `gpt-oss-120b`'nin günlük token kotası dolunca aynı ailenin küçük modelini kullandım.

**2. En işe yarayan iki istem**

*(a) Claude'a:* "km neden analizde kullanılmayacak olarak değerlendirdin?"
Bu soru, gerekçesiz bir varsayımı ortaya çıkardı ve modele güçlü bir değişken eklenmesini sağladı (3. maddede).

*(b) LLM'e, aksiyon önerisi için (kısaltılmış):*
"Temsilcinin yapabilecekleri: müşteriyi aramak, özür dilemek, bilgilendirmek, şikâyeti ilgili birime iletmek... Şirketin süreçlerini değiştiremez. Hiçbir kampanya, indirim, iade, hediye veya ücretsiz hizmet uydurma; oran veya rakam yazma. Teklif gerekiyorsa temsilcinin şirket sistemindeki güncel kampanyaları kontrol etmesini söyle."
Kısıtsız prompt "%10 indirimli paket", "indirim kuponu" gibi taahhütler uydurmuştu; bu prompt ve çıktıyı kodla denetleyen kontrol ile uydurma ortadan kalktı.

**3.Aracın yanıldığı örnek**

Veri kalitesi kontrolünde Claude, km sütununda negatif değerler ve gerçekçi olmayan bir ölçek tespit etti ve "km analizde kullanılmayacağı için yalnızca not ediliyor" dedi. Bu bir analiz kararı değil, gerekçesiz bir varsayımdı: Bir sütunda kalite sorunu olması, bilgi taşımadığı anlamına gelmiyor.
*Nasıl fark ettim:* Neden kullanılmayacağını sorduğumda bunun bilinçli bir karar olmadığı ortaya çıktı. Kontrol edince km'nin tutarlı olduğu (araç başına zamanla artıyor, negatifler işaret hatası) ve aynı yaştaki araçlarda güçlü bir terk sinyali taşıdığı görüldü (9+ yaş: düşük km'de terk %41, yüksek km'de %17). km modele eklendi ve en güçlü değişkenlerden biri oldu.

Aynı şekilde sonucu değiştiren diğer sorgulamalarım: Model farklarının çok küçük olduğunu fark edip daha karmaşık bir model aramak yerine modeli 22'den 6 değişkene sadeleştirdik; etiketleme kılavuzunun LLM'e olduğu gibi verilmesine itiraz edip bağımsız bir prompt yazdırdım; 3.3'te sabit bir aksiyon kataloğunun öngörülmeyen şikâyetlerde işe yaramayacağını düşünüp yalnızca parasal taahhütleri kısıtlayan bir tasarıma geçtik. Etiketleyen LLM ise ironiyi ve çift olumsuzu kaçırdı; ironi hatası yüzünden en yüksek riskli şikâyetçi müşteri aksiyon listesine girmedi.

**4. Bilinçli olarak devretmediğim iş**
**Altın setin etiketlenmesi.** LLM'i ölçmek için ondan bağımsız bir referans gerekiyordu; etiketleri bir yapay zekâya yaptırsaydım, iki model aynı hatayı yaptığında karşılaştırma bunu doğru sayardı. Etiketlememde, etiketleme kılavuzu oluşturdum ve 35 metni kılavuza göre etiketledim.
